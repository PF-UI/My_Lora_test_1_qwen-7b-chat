```python
import json
import os

file_path = "weiwuxian_dialogues_dataset.json"

if not os.path.exists(file_path):
    print(f"文件不存在：{file_path}")
else:
    with open(file_path, "r", encoding="utf-8") as f:
        dataset = json.load(f)
    
    # 第一层：查看chapters键对应的值
    chapters = dataset.get("chapters")
    print(f"1. 'chapters'的值类型：{type(chapters)}")
    
    if isinstance(chapters, list):
        print(f"   'chapters'是列表，包含 {len(chapters)} 个元素（章节）")
        
        # 第二层：查看第一个章节的结构
        if len(chapters) > 0:
            first_chapter = chapters[0]
            print(f"\n2. 第一个章节的类型：{type(first_chapter)}")
            
            if isinstance(first_chapter, dict):
                print(f"   第一个章节包含的键：{list(first_chapter.keys())}")
                
                # 第三层：查看第一个章节内的关键字段（示例：假设可能有'dialogues'等键，按需调整）
                # 先打印第一个章节的完整结构（格式化）
                print("\n3. 第一个章节的详细结构：")
                print(json.dumps(first_chapter, ensure_ascii=False, indent=2))
                
                # （可选）如果章节内有嵌套列表（如对话列表），可进一步查看
                # 例如，若有'dialogues'键，查看其结构：
                if "dialogues" in first_chapter:
                    dialogues = first_chapter["dialogues"]
                    print(f"\n4. 第一个章节中'dialogues'的类型：{type(dialogues)}")
                    if isinstance(dialogues, list) and len(dialogues) > 0:
                        print(f"   'dialogues'包含 {len(dialogues)} 条对话")
                        print("   第一条对话的结构：")
                        print(json.dumps(dialogues[0], ensure_ascii=False, indent=2))
            else:
                print("   第一个章节不是字典（不常见，可能为字符串/其他类型）")
        else:
            print("   'chapters'列表为空，无内容")
    
    elif isinstance(chapters, dict):
        print(f"   'chapters'是字典，包含的键：{list(chapters.keys())}")
        # 若chapters是字典，可进一步查看其内部键（例如按章节ID划分）
        first_key = next(iter(chapters.keys())) if chapters else None
        if first_key:
            print(f"\n2. 'chapters'中第一个键'{first_key}'对应的值结构：")
            print(json.dumps(chapters[first_key], ensure_ascii=False, indent=2))
    else:
        print("   'chapters'的值既不是列表也不是字典（不常见格式）")
```
